```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Function, f(x)
# Derivitive of Function, f'(x)
# initialisation of parameters (starting point x, learning rate (alpha), max_iterations, tolerance )
# gradient ascent (return number of iterations, x values, f(x) values, slop/gradient values, option to plot)
# if plot == True, plot x_values vs iterations, f(x) vs iterations, gradient vs iterations, title = plot with alpha (learning rate)
```

# Gradient Ascent

```python
def f(x):
  return -(x-3) **2 + 10

def d_f(x):
  return -2*(x-3)
```

```python
def gradient_ascent(f_x, d_f, initial_x, alpha, tolerance, max_iterations, is_plot = False):
  x = initial_x
  tol = tolerance
  max_iter = max_iterations

  x_values = []
  f_x_values = []
  gradient_values = []

  iterations_value = []

  for i in range(0,max_iter,1):
    gradient = d_f(x)

    new_x = x + alpha *gradient


  # stores values for plotting
    x_values.append(x)
    f_x_values.append(f_x(x))
    gradient_values.append(gradient)
    iterations_value.append(i)

    if abs(f_x(new_x) - f_x(x)) < tol:
      break
    else:
      x = new_x




  if is_plot == True:
    plt.figure(figsize = (10,10))
    plt.title(f"Gradient Ascent with Alpha = {alpha}")

    plt.plot(iterations_value, x_values, color = "red", label = "x")
    plt.plot(iterations_value, f_x_values,color = "blue", label = "f(x)")
    plt.plot(iterations_value, gradient_values ,color = "green", label = "Gradient")
    plt.plot(iterations_value[0], initial_x, "ro", label = f"Initial Point at x = {initial_x}")
    plt.xlabel("Iterations")
    plt.ylabel("Values")


    plt.grid(True)

    plt.legend()

    plt.legend()
    plt.show()

  return x_values, f_x_values, gradient_values, iterations_value

```

```python
x_values, f_x_values, gradient_values, iterations_value = gradient_ascent(f_x = f, d_f = d_f, initial_x = 5, alpha=0.2, tolerance = 1e-6, max_iterations=1_000, is_plot = False)
```

```python

gradient_ascent(f_x = f, d_f = d_f, initial_x = 5, alpha=0.2, tolerance = 1e-6, max_iterations=10, is_plot = True)
```

# Gradient Descent

```python
def f_min(x):
  return x**2

def d_f_min(x):
  return 2*x
```

```python
def gradient_descent(f_x, d_f, initial_x, alpha, tolerance, max_iterations, is_plot = False):
  x = initial_x
  tol = tolerance
  max_iter = max_iterations

  x_values = []
  f_x_values = []
  gradient_values = []

  iterations_value = []

  for i in range(0,max_iter,1):
    gradient = d_f(x)

    new_x = x - alpha *gradient


  # stores values for plotting
    x_values.append(x)
    f_x_values.append(f_x(x))
    gradient_values.append(gradient)
    iterations_value.append(i)

    if abs(f_x(new_x) - f_x(x)) < tol:
      break
    else:
      x = new_x




  if is_plot == True:
    plt.figure(figsize = (10,10))
    plt.title(f"Gradient Descent with Alpha = {alpha}")

    plt.plot(iterations_value, x_values, color = "red", label = "x")
    plt.plot(iterations_value, f_x_values,color = "blue", label = "f(x)")
    plt.plot(iterations_value, gradient_values ,color = "green", label = "Gradient")
    plt.plot(iterations_value[0], initial_x, "ro", label = f"Initial Point at x = {initial_x}")
    plt.xlabel("Iterations")
    plt.ylabel("Values")


    plt.grid(True)


    plt.legend()
    plt.show()

  return x_values, f_x_values, gradient_values, iterations_value

```

```python
x_values, f_x_values, gradient_values, iterations_value = gradient_descent(f_x = f_min, d_f = d_f_min, initial_x = 10, alpha=0.7, tolerance = 1e-6, max_iterations=1_000, is_plot = True)
```

```python
print(x_values, f_x_values, gradient_values, iterations_value )
```

```python
def f_2d(vector):
  x = vector[0]
  y = vector[1]

  return (x-3)**2 +(y+2)**2

def grad_f_2d(vector):
  x = vector[0]
  y = vector[1]

  return np.array([
      2*(x-3),
      2*(y+2)
  ])
```

```python
initial_x_2d = np.array([10,10])
```

```python
def gradient_descent_2d(f_x, d_f, initial_x, alpha, tolerance, max_iterations, is_plot = False):
  x = initial_x
  tol = tolerance
  max_iter = max_iterations

  x_values = []
  f_x_values = []
  gradient_values = []

  iterations_value = []

  for i in range(0,max_iter,1):
    gradient = d_f(x)

    new_x = x - alpha *gradient


  # stores values for plotting
    x_values.append(x.copy())
    f_x_values.append(f_x(x))
    gradient_values.append(gradient.copy())
    iterations_value.append(i)

    if abs(f_x(new_x) - f_x(x)) < tol:
      break
    else:
      x = new_x




  if is_plot == True:
    plt.figure(figsize = (10,10))
    plt.title(f"2D Gradient Descent with Alpha = {alpha}")

    x_coords = [point[0] for point in x_values]
    y_coords = [point[1] for point in x_values]
    grad_x_vals = [grad[0] for grad in gradient_values]
    grad_y_vals = [grad[1] for grad in gradient_values]

    plt.plot(iterations_value, x_coords, label = "x")
    plt.plot(iterations_value, y_coords,  label = "y")
    #plt.plot(iterations_value, f_x_values, label = "f(x,y)")
    plt.plot(iterations_value, grad_x_vals , label = "Gradient, x_component")
    plt.plot(iterations_value, grad_y_vals, label = "Gradient, y_component")
    plt.plot(0, initial_x[0], "bo", label = f"Initial x_coord: {initial_x[0]}")
    plt.plot(0, initial_x[1], "bo", label = f"Initial y_coord {initial_x[1]}")
    plt.xlabel("Iterations")
    plt.ylabel("Values")


    plt.grid(True)


    plt.legend()
    plt.show()

  return x_values, f_x_values, gradient_values, iterations_value

```

```python
x_values, f_x_values, gradient_values, iterations_value = gradient_descent_2d(f_2d, grad_f_2d, initial_x_2d, alpha = 0.05, tolerance=1e-6, max_iterations=1_000, is_plot = True)
```
